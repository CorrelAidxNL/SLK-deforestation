```python
# unzip all files in the directory

os.chdir(path)
# Unzip all archives (creates the .tif files)
for f in os.listdir(path):
    if f.endswith(".zip"):
        with zipfile.ZipFile(f, 'r') as zip_ref:
            zip_ref.extractall(path)
```

```python
# Merge all TIF files into one
!gdal_merge.py -o {merged_file_name} -of GTiff -co COMPRESS=LZW *.tif
```

```python
#clip tiff by country boundary

command = [
    "gdalwarp",
    "-overwrite",
    "-s_srs", "EPSG:3035",
    "-of", "GTiff",
    "-cutline", country_boundary,
    "-cl", country_boundary_layer,
    "-crop_to_cutline",
    merged_file_path,
    clipped_file_path
]

subprocess.run(command, check=True)
```

```python
#replace 0 with Undefined
command = [
    "gdal_translate",
    "-a_nodata", "0",
    clipped_file_path,
    clipped_eliminated_file_name
]

subprocess.run(command, check=True)
```
